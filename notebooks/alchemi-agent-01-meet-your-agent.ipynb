<h2 style="display: block; color: inherit; background: transparent; border: 1px solid rgba(127,127,127,0.35); border-left: 4px solid #76b900; border-radius: 4px; padding: 15px 20px; margin: 24px 0 14px; font-family: NVIDIA Sans, Helvetica Neue, Arial, sans-serif; font-size: 20px; font-weight: 500;">1. Work with your agent</h2>

You direct a coding agent, and the agent writes the Toolkit code.

- **You write the prompt.** This is the most important step: it decides
  what will be done.
- The agent writes a short Python file and checks it.
- You load the file with `%load`, read it, and run it.
- A table or plot shows the result, and one line tells you what to look for.

<img src="assets/agent01-loop-light.svg" alt="A cycle: prompt, file, load and run, result; the agent writes the file." style="display:block;box-sizing:border-box;width:100%;max-width:920px;height:auto;margin:0;">

<h2 style="display: block; color: inherit; background: transparent; border: 1px solid rgba(127,127,127,0.35); border-left: 4px solid #76b900; border-radius: 4px; padding: 15px 20px; margin: 24px 0 14px; font-family: NVIDIA Sans, Helvetica Neue, Arial, sans-serif; font-size: 20px; font-weight: 500;">2. What a good prompt holds</h2>

| a good prompt holds | for example |
| --- | --- |
| the skill: here the nvalchemi-toolkit skill; other tools have their own | `use the nvalchemi-toolkit skill` |
| the task, in one line | `compute the energy of one water molecule` |
| the input: data and models, by their names in the notebook | `model: AIMNet2` |
| the output: the names you want back | `give me water_energy` |
| optional, for fine control: name an exact setting, or ask the agent why it chose one | `tell me which settings you chose` |

Put together:

```text
use the nvalchemi-toolkit skill
- compute the energy of one water molecule with AIMNet2
- give me water_energy
```
